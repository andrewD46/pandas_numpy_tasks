* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
# python implementation
x = [
        [1, 0, 1],
        [2, 0, 2],
        [3, 0, 3],
        [4, 4, 4]
]
mul = 1
for i in range(len(x[0])):
    mul *= x[i][i] if x[i][i] != 0 else 1

mul
```

```python
# numpy
x = np.array(
    [
        [1, 0, 1],
        [2, 0, 2],
        [3, 0, 3],
        [4, 4, 4]
])
mul = np.prod(np.diag(x)[np.diag(x) != 0])
mul
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
# python
x = [2, 2, 2, 3, 3, 3, 5]
x = sorted(x)
nums = []
count = []
n = x[0]
c = 0
for i in x:
    if i == n:
        c += 1
    else:
        nums.append(n)
        count.append(c)
        n = i
        c = 1
nums.append(n)
count.append(c)

print(nums, count)
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import random
from scipy.spatial.distance import cdist
```

```python
%time
random.seed(123)
n, m, dims = 100, 100, 10
X = [[random.random() for _ in range(dims)] for _ in range(n)]
Y = [[random.random() for _ in range(dims)] for _ in range(m)]

res = [[0 for _ in range(m)] for _ in range(n)]

for i in range(n):
    for j in range(m):
        s = sum((x - y)**2 for x, y in zip(X[i], Y[j]))
        res[i][j] = s**0.5

res[0][0]
```

```python
%time
# fancy func
X = np.array(X)
Y = np.array(Y)
res = cdist(X, Y, metric='euclidean')
res
```
